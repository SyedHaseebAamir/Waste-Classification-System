```python
import os
import shutil

def sort_dataset(images_dir,labels_dir,c_0_img,c_0_lab,c_1_img,c_1_lab):

    os.makedirs(c_0_img, exist_ok=True)
    os.makedirs(c_0_lab, exist_ok=True)
    os.makedirs(c_1_img, exist_ok=True)
    os.makedirs(c_1_lab, exist_ok=True)

    for filename in os.listdir(images_dir):
        if filename.endswith(".jpg"):  # Assuming images are JPEG format
            image_path = os.path.join(images_dir, filename)
            label_path = os.path.join(labels_dir, filename[:-4] + ".txt")

            # Read the annotation file
            with open(label_path, 'r') as file:
                annotations = file.readlines()

            # Check the class in annotations and move the image and label accordingly
            for annotation in annotations:
                class_id = int(annotation.split()[0])
                if class_id == 0:
                    shutil.copy(image_path, os.path.join(c_0_img, filename))
                    shutil.copy(label_path, os.path.join(c_0_lab, filename[:-4] + ".txt"))
                elif class_id == 1:
                    # Change class id to 6
                    annotation = '6' + annotation[1:]
                    # Save modified annotation to a new file
                    new_label_path = os.path.join(c_1_lab, filename[:-4] + ".txt")
                    with open(new_label_path, 'a') as new_file:
                        new_file.write(annotation)
                    shutil.copy(image_path, os.path.join(c_1_img, filename))


sort_dataset(
    images_dir=r"/content/drive/MyDrive/last/dry-and-wet-waste-1/train/images",
    labels_dir=r"/content/drive/MyDrive/last/dry-and-wet-waste-1/train/labels",
    c_0_img=r"/content/drive/MyDrive/last/dry-and-wet-waste-1/class_0/images",
    c_0_lab=r"/content/drive/MyDrive/last/dry-and-wet-waste-1/class_0/labels",
    c_1_img=r"/content/drive/MyDrive/last/dry-and-wet-waste-1/train_1/images",
    c_1_lab=r"/content/drive/MyDrive/last/dry-and-wet-waste-1/train_1/labels"
)

sort_dataset(
    images_dir=r"/content/drive/MyDrive/last/dry-and-wet-waste-1/valid/images",
    labels_dir=r"/content/drive/MyDrive/last/dry-and-wet-waste-1/valid/labels",
    c_0_img=r"/content/drive/MyDrive/last/dry-and-wet-waste-1/class_0/images",
    c_0_lab=r"/content/drive/MyDrive/last/dry-and-wet-waste-1/class_0/labels",
    c_1_img=r"/content/drive/MyDrive/last/dry-and-wet-waste-1/valid_1/images",
    c_1_lab=r"/content/drive/MyDrive/last/dry-and-wet-waste-1/valid_1/labels"
)

```

```python
from pathlib import Path
import yaml
from collections import Counter

# Paths
label_dirs = [
    Path("/content/drive/MyDrive/last/Waste-Segregation-7/train/labels"),
    Path("/content/drive/MyDrive/last/Waste-Segregation-7/valid/labels"),
    Path("/content/drive/MyDrive/last/Waste-Segregation-7/test/labels"),
]

yaml_path = "/content/drive/MyDrive/last/Waste-Segregation-7/data.yaml"

# Load class names from data.yaml
with open(yaml_path, 'r') as f:
    data_yaml = yaml.safe_load(f)

names = data_yaml.get("names", [])
nc = data_yaml.get("nc", len(names))

# Count instances per class
class_counts = Counter()

for label_dir in label_dirs:
    for file in label_dir.glob("*.txt"):
        with open(file) as f:
            for line in f:
                parts = line.strip().split()
                if parts:
                    class_id = int(parts[0])
                    class_counts[class_id] += 1

# Display results
print("✅ Class usage across dataset:\n")
for class_id in sorted(class_counts.keys()):
    name = names[class_id] if class_id < len(names) else "(Unnamed)"
    print(f"Class {class_id:2}: {name:20} → {class_counts[class_id]} objects")
```

```python
with open("/content/drive/MyDrive/app.py", "w") as f:
    f.write("""# Streamlit app for Waste Detection using YOLOv5
import os
import cv2
import torch
import shutil
import streamlit as st
from pathlib import Path
from PIL import Image

# ==== Configurable Paths ====
MODEL_PATH = '/content/drive/MyDrive/last/yolov5'
MODEL_FILE_PATH = '/content/drive/MyDrive/last/yolov5/waste-segmentation-run/exp1/weights/best.pt'
IMG_SAVE_PATH = 'runs/detect/exp/annotated.jpg'
TEMP_DIR = 'runs/detect/exp'

os.makedirs(TEMP_DIR, exist_ok=True)

class_colors = {
    'cardboard': (0, 255, 0),
    'plastic bottle': (255, 0, 0),
    'paper': (0, 0, 255),
    'can': (0, 255, 255),
    'tissue': (255, 255, 0),
    'wet waste': (128, 0, 128),
    'plastic cup': (255, 102, 255),
    'glass bottle': (102, 255, 255),
    'metal': (100, 100, 100),
    'thermocol': (255, 204, 0),
    'junk food wrapper': (255, 153, 153),
    'napkin': (153, 255, 153),
    'diaper': (255, 102, 102),
    'crumpled paper': (204, 204, 255),
    'leaves': (0, 128, 0)
}

recyclable_map = {
    'cardboard': 'Recyclable, Paper',
    'plastic bottle': 'Recyclable, PET',
    'glass bottle': 'Recyclable, Glass',
    'can': 'Recyclable, Metal',
    'metal': 'Recyclable, Metal',
    'paper': 'Recyclable, Paper',
    'plastic cup': 'Recyclable, Plastic',
    'crumpled paper': 'Recyclable, Paper',
    'tissue': 'Non-Recyclable, Tissue',
    'napkin': 'Non-Recyclable, Paper',
    'wet waste': 'Non-Recyclable, Organic',
    'diaper': 'Non-Recyclable, Disposable',
    'thermocol': 'Non-Recyclable, Foam',
    'junk food wrapper': 'Non-Recyclable, Mixed',
    'leaves': 'Compostable, Organic'
}

def load_model():
    return torch.hub.load(MODEL_PATH, 'custom', path=MODEL_FILE_PATH, source='local')

def draw_predictions(image_path):
    model = load_model()
    img = cv2.imread(image_path)
    results = model(img)
    summary = []
    for _, row in results.pandas().xyxy[0].iterrows():
        x1, y1, x2, y2, conf, cls, name = row
        color = class_colors.get(name, (0, 0, 255))
        category = recyclable_map.get(name.lower(), 'Unknown')
        label_text = f'{name} ({category}) {conf:.2f}'
        summary.append(label_text)
        cv2.rectangle(img, (int(x1), int(y1)), (int(x2), int(y2)), color, 2)
        cv2.putText(img, label_text, (int(x1), int(y1)-10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 2)
    out_path = os.path.join(TEMP_DIR, 'annotated.jpg')
    cv2.imwrite(out_path, img)
    return out_path, summary

# ==== Streamlit Interface ====
st.set_page_config(page_title="Waste Detection", layout="centered")

st.title("♻️ Waste Object Detection")

uploaded_file = st.file_uploader("Upload an image", type=['jpg', 'jpeg', 'png'])

if uploaded_file is not None:
    temp_path = os.path.join(TEMP_DIR, uploaded_file.name)
    with open(temp_path, "wb") as f:
        f.write(uploaded_file.getbuffer())

    st.image(temp_path, caption="Uploaded Image", use_column_width=True)

    with st.spinner("Detecting objects..."):
        result_img, summary = draw_predictions(temp_path)
        st.image(result_img, caption="Prediction Result", use_column_width=True)
        st.markdown("### 🧾 Detected Waste Items:")
        for item in summary:
            st.markdown(f"- {item}")

        with open(result_img, "rb") as f:
            st.download_button("📥 Download Annotated Image", f, file_name="annotated_result.jpg")
# Display average confidence score
results = load_model()(cv2.imread(temp_path))
confidences = [row['confidence'] for _, row in results.pandas().xyxy[0].iterrows()]
if confidences:
    avg_conf = sum(confidences) / len(confidences)
    st.success(f"📈 Average Confidence: {avg_conf:.2f}")
else:
    st.warning("No objects detected, so no confidence score available.")

""")
```

```python
# Force inline plotting
%matplotlib inline
import matplotlib.pyplot as plt
from collections import Counter

# Run detection
import torch
model = torch.hub.load('yolov5', 'custom', path='/content/drive/MyDrive/last/yolov5/waste-segmentation-run/exp1/weights/best.pt', source='local')
img = '/content/drive/MyDrive/bin.png'
results = model(img)

# Parse results
df = results.pandas().xyxy[0]

# Display results
if df.empty:
    print("⚠️ No objects detected. Please try with a different image.")
else:
    print(df[['name', 'confidence']])

    # Plot graph
    detected_classes = df['name'].tolist()
    class_counts = Counter(detected_classes)

    plt.figure(figsize=(10, 5))
    plt.bar(class_counts.keys(), class_counts.values(), color='skyblue')
    plt.title("🗑️ Detected Waste Type Distribution")
    plt.xlabel("Waste Category")
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```

```python
import os
from collections import Counter
import matplotlib.pyplot as plt
from pathlib import Path
import yaml

# === Step 1: Load YAML to get class names ===
yaml_path = "/content/drive/MyDrive/last/Waste-Segregation-7/data.yaml"
with open(yaml_path, "r") as f:
    data = yaml.safe_load(f)
names = data["names"]

# === Step 2: Label directory path ===
label_dir = "/content/drive/MyDrive/last/Waste-Segregation-7/train/labels"


# === Step 3: Parse all .txt files ===
all_labels = []
for label_file in Path(label_dir).glob("*.txt"):
    with open(label_file, "r") as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) > 0:
                class_id = int(parts[0])
                all_labels.append(class_id)

# === Step 4: Count and plot ===
if not all_labels:
    print("⚠️ No class labels found. Check label format or path.")
else:
    class_names = [names[i] for i in all_labels]
    counts = Counter(class_names)

    plt.figure(figsize=(12, 6))
    plt.bar(counts.keys(), counts.values(), color="mediumseagreen")
    plt.title("🗂️ Dataset Class Distribution")
    plt.xlabel("Class Name")
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.grid(axis="y", linestyle="--", alpha=0.7)
    plt.tight_layout()
    plt.show()
```
